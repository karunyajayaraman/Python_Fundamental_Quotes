#Case Study: Sales Tracking System

A retail store records sales transactions every day. The manager wants a Python function that can:

Accept any number of sales amounts.
Calculate the current day's total sales.
Maintain a running total of all sales made till date.
Display both the current sale total and the cumulative sales total.

Task:
Write a Python function total_sale(*sales) that uses a global variable to keep track of the total sales till date.

```python
total_sale_till = 0 #initializes the cumulative sales.

def total_sales(*sales):   #collects all the sales values into a tuple.
  global total_sale_till  # Declared the Global variable  #allows the function to modify the global variable.

  current_total_sale = sum(sales)  # Local variable  # calculates the current day's sales.
  total_sale_till += current_total_sale

  #Each time total_sale() is called:

  #It calculates the current day's sales.
  #Adds it to the cumulative total.
  #Prints both values.

  print(f"Current total sales till date :  {current_total_sale}")
  print(f"Total Sales : {total_sale_till}")

total_sales(200,400,500,650)

total_sales(1200,1500,1350,2500,1100,900,3500)
```
