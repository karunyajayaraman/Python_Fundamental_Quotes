#Case Study – Employee Salary Analysis
Instructions

Create a list called salary Store employee salaries inside the list

Use Built-in Functions to:

Find maximum salary

Find minimum salary

Find total salary

Find number of employees using len()

Use lambda with map() to: Increase each salary by 10%

Use lambda with filter() to: Display salaries greater than 50000

Display:

Original salary list

Updated salary list

High salary list

Maximum salary

Minimum salary

Total salary

```python
# Employee Salary Analysis

#Create a list
employee_salary = [35000,25000,45000,15000,50000]

#Use Built-in Functions

max_salary = max(employee_salary)
min_salary = min(employee_salary)
sum_salary = sum(employee_salary)
length_emp = len(employee_salary)

#Increase each salary by 10%
# use lambda

disc_sal = list(map(lambda e : e + (e * 1.10),employee_salary))

#Display salaries greater than 50000
#use filter -> disc_sal( [73500.0, 52500.0, 94500.0, 136500.0, 105000.0] )

greater_sal = list(filter(lambda e : e > 50000, disc_sal))

#high_salary = list(filter(lambda x: x > 50000, updated_salary))


print(f"****Employee Salary Analysis****")

print(f"\nEmployee Original Salary : ",employee_salary)
print(f"\nMaximum Salary : ",max_salary)
print(f"Manimum Salary : ",min_salary)
print(f"Total Employees Salary : ",sum_salary)
print(f"Number of employees : ",length_emp)
print(f"Increase the salary by 10% : ",disc_sal)
print(f"Greater than 50000: ",greater_sal)
```
