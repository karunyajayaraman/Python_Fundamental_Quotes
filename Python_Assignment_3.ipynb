```python
#Tasks:
#Task 1: While Loop & Control Statements -Number Guessing Game

#1. Set Up the Game:

#★ Generate a random number between 1 and 10 that the user has to guess.
#Import random and use randint function.


print(f'Welcome to the Number Guessing Game')

import random # Generate random number between 1 and 10
secret_number = random.randint(1, 10)
print(secret_number)
```

```python
import random
secret_number = random.randint(1, 10)

attempts = 0  # Attempt variables
max_attempts = 3

print("Welcome to the Number Guessing Game!")

while attempts < max_attempts:

    user_guess = int(input("\nGuess the number (between 1 and 10): "))


    if user_guess < 1 or user_guess > 10:  ## Check the guess
        print("Your guess is out of range. Please enter a number between 1 and 10.")
        continue

    attempts += 1  # increamented by 1

    if user_guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break

    elif user_guess > secret_number:
        print("Too high. Try again.")

    else:
        print("Too low. Try again.")

    print(f"Attempts left: {max_attempts - attempts}")

else:
    print("\nBetter luck next time!")
    print(f"The correct number was {secret_number}.")
```

```python
#Task 2:- For Loop - Multiplication Table Generator

mul_number = int(input("Enter the number for which you want the multiplication table : "))

print(f'\nMultiplication table of {mul_number}\n')

for i in range(1,11):
  result = mul_number * i
  print(f'{mul_number} * {i} = {result}')
```

```python
#Task 3: Function- BMI Calculator

#Hint: BMI = weight (kg) / [height (m)]²

def calculate_bmi(weight, height):
  bmi = round(weight / (height**2))  # if height in cm - convert it to meter
  return bmi

#Get input from user for their weight (in kg) and height (in meters)

weight = float(input("Enter your Weigh(kg) : "))
height = float(input("Enter your Height(m) : "))
bmi = calculate_bmi(weight, height)

print(f'Your BMI is : {bmi}')
```
