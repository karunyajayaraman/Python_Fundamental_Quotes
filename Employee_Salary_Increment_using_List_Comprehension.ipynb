#Case Study – Employee Salary Increment using List Comprehension
Instructions

Create a list called salary

Store employee salaries inside the list

Use list comprehension to: Increase each salary by 10%

Create another list to: Display only salaries greater than 50000

Print:

Original salary list

Updated salary list

Filtered salary list

```python
salary = [45000,55000,70000,35000]

#Increase each salary by 10%

#salary = salary *  10%

increased_sal = [(round(s * 01.10)) for s in salary]
#print(f'Increased Salary by 10% :' , increased_sal)

#Display only salaries greater than 50000
# Increased Salary by 10% : [49500, 60500, 77000, 38500]

high_salary = [i for i in increased_sal if i >= 50000]
#print(f"Salary Greater than 50000 : ", high_salary)

print(f'Original salary list : ' ,salary)

print(f'Updated salary list : ',increased_sal)

print(f'Filtered salary list : ',high_salary)
```
