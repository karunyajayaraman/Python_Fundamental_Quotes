Numpy Array Operations:

```python
import numpy as np
```

```python
# 1. Create a 1D NumPy Array:

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9], dtype = float)

print(f'Week1_Temperatures :',temperatures_w1)
temperatures_w1
```

```python
# 2. Inspection and Properties:

print(f'Shape of Temperatures_W1 : ',(temperatures_w1).shape)
print(f'DataType : ',(temperatures_w1).dtype)
print(f'Number of elements in the array : ',(temperatures_w1).size)
```

```python
# 3. Array Operations:

#Celsius to Fahrenheit - (Fahrenheit = (Celsius * 9/5) + 32)

temperatures_w1

Fahrenheit_w1 = ((temperatures_w1 * 9/5)+32)

print('Fahrenheit of Week1 :',Fahrenheit_w1)
print('Temperature of Week1 :',temperatures_w1)

print('\n*****Temperatures for the week*****')
print('\nMinimum :',temperatures_w1.min())
print('Maximum :',temperatures_w1.max())
print('Mean :',round(temperatures_w1.mean(),2))   # Mean : 74.37714285714286
```

```python
# 4. Array Slicing and Indexing:

print("Temperatures of the Week")
print('\nFirst three days of the week :',temperatures_w1[:3])
print('Weekend (last two days) :',temperatures_w1[5:])     # [-2:]  also in negative indexing
print('Middle Three days of the week :',temperatures_w1[2:5])
```

```python
# 5. Create a 2D Array:

#Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)

#Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] ,
                        [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]]
                        ,dtype = float)
temperatures

print("Temperatures of Week 1 & 2 : ",temperatures)
```

```python
# 6. Inspect and Slice the 2D Array:

print(f'Shape of Temperatures : {(temperatures).shape}')
print(f'DataType : {(temperatures).dtype}')
print(f'Number of elements in the array : {(temperatures).size}')

print(f'\nWeek 1 - Temperature : {temperatures[0]}')
print(f'Week 2 - Temperature : {temperatures[1]}')

print(f'\nWeekends Temperature(Last 2 days):\n{temperatures[0:,5:]}')
```

#Pandas Series:


```python
#Importing pandas
import pandas as pd
```

```python
# 1. Creating Pandas Series:

marks = [95, 92, 89, 85, 80]
ranks = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']

scores = pd.Series(marks, index= ranks)
print(scores)
scores
```

```python
# 2. Indexing and Slicing:

# For the Pandas Series 'scores':

print(f'\nMark of the 1st rank student (from Series) : {scores.iloc[0]}')  # Integer Index Position(pandas)

print(f'\nMarks of the Top 3 ranks (from Series): \n{scores.loc['Rank1':'Rank3']}')  #Rank by Label using loc

print(f'\nMark of the 3rd-ranked student : \n {scores.iloc[2]}') #iloc - integer location

print(f'\nMarks are greater than 90 : \n{scores.loc[scores > 90]}')

```

```python
# 3. Manipulating Series:
scores
```

```python
# Modify - 1st rank to 100.

scores.loc['Rank1'] = 100
scores

print(f'\nModified the 1st rank(100) : \n{scores}')
```

```python
# Remove the Last Rank from the Series - use DROP

# scores.drop('Rank5')  # Direct method
# scores.index[-1] - select the last rank from the series(way to used for large set of datas)

print(f'\nRemove the Last rank :\n{scores.drop(scores.index[-1])}')
```

```python
# CGPA - Divide the mark in the Series by 10

cgpa = scores / 10  # Vectorized operations

print(f'\nCGPA :\n{cgpa}')
```

#Pandas DataFrame:

```python
# Dictionary

transactions = {'TransactionID' : [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
                'ProductCategory':['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
                'Region' : ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
                'Amount' : [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
              }
transactions
```

```python
# 1. Creating Pandas DataFrame:

transactions = pd.DataFrame(transactions)
transactions
```

```python
# 2. Data Exploration:

#Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.

# Head()

print("******Basic Information*****")

print(f'\nFirst 5 Rows \n')

transactions.head()
```

```python
#Tail()

print(f'\nLast 5 Rows \n')

transactions.tail()
```

```python
#Shape

print(f'\nShape of Transactions : {transactions.shape}')
```

```python
# Column Names:

print(f'\nColumn Name of Transactions :\n\n{transactions.columns}')
```

```python
# Find the unique values in the 'Region' column

print(f'Unique values from Region : {transactions["Region"].unique()}')
```

```python
# Group by 'Region' and find the mean amount for each region

transactions.groupby('Region')['Amount'].mean().reset_index()
```
