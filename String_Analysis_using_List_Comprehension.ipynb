#Case Study – String Analysis using List Comprehension
**Instructions **

Create a string variable called text

Store any sentence or word inside the variable

Use list comprehension to:

Convert all characters into uppercase

Extract only vowels from the string

Extract only consonants from the string

Display:

Original string

Uppercase characters

Vowels list

Consonants list

String without spaces

Print the output in a clean format using print() statements

```python
text = "Data analytics"

#Convert all characters into uppercase
upper_case = [t.upper() for t in text]


#Extract only vowels from the string
vowels = [t for t in text if t in 'aeiou']


#Extract only consonants from the string
#consonants = [t for t in text if t not in 'aeiou']
consonants = [t for t in text if t.isalpha() and t not in 'aeiou']

#Removed spaces
remove_spaces = [x for x in text if x != " "]

print(f'Text = Data analytics')

print(f'\nUpper Case   :{upper_case}')
print(f'Vowels         : {vowels}')
print(f'Consonants     : {consonants}')
print(f'Removed Spaces : {remove_spaces}')
```
